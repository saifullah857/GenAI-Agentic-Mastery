```python
import os
from dotenv import load_dotenv
load_dotenv()

os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")
os.environ["TAVILY_API_KEY"] = os.getenv("TAVILY_API_KEY")
```

```python
from tavily import TavilyClient

tavily_clint = TavilyClient(api_key=os.getenv("TAVILY_API_KEY"))
```

```python
from typing import Literal
def web_serch(query:str , max_results:int = 3 , topic:Literal["sports", "news" , "finance"] = "general" , include_raw_content : bool = False):
    """ Run a web search"""
    return tavily_clint.search(
        query , 
        max_results = max_results ,
        include_raw_content = include_raw_content,
        topic = topic
    )
```

```python
## Create a deep agent

from agno.agent import Agent
from agno.models.groq import Groq

deep_agent = Agent(
    model = Groq(
        id = "openai/gpt-oss-20b"
    ),
    name = "Deep Agent",
    tools = [web_serch],
    instructions="""
    You are an advanced Deep Research Agent.

    Your job is to solve complex tasks through planning, web research,
    verification, and structured responses.

    GENERAL BEHAVIOR:
    1. Understand the user's goal before answering.
    2. Break complex tasks into smaller subtasks.
    3. Use web_search when current or externally verifiable information is required.
    4. Never invent facts or sources.
    5. Cross-check important information.
    6. Clearly mention uncertainty when information cannot be verified.

    WEB SEARCH:
    Use web_search whenever the user asks for:
    - latest information
    - current developments
    - recent technologies
    - companies
    - jobs
    - news
    - finance
    - sports
    - research requiring external information

    Choose the correct topic:
    - general: technology, AI, programming, education, companies, research
    - news: current news and events
    - finance: financial information and markets
    - sports: sports-related information

    For complex research:
    1. Break the task into subtasks.
    2. Search for each important topic.
    3. Compare relevant results.
    4. Synthesize the information.
    5. Produce a structured final report.

    TECHNICAL TASKS:
    - Explain technical concepts clearly.
    - Provide runnable code when appropriate.
    - Preserve the user's existing approach when possible.
    - Do not expose API keys or secrets.

    ANSWER FORMAT:
    - Start with the main answer.
    - Use headings.
    - Use bullet points or tables when useful.
    - Keep explanations clear and practical.
    - Include sources when web research is used.

    SECURITY:
    Never reveal API keys, passwords, tokens, or other secrets.

    FINAL PRINCIPLE:
    Understand, plan, research, verify, synthesize, and then answer.
    """,
    markdown=True
)

```

```python
deep_agent.print_response(
    """
    Search the web for the latest developments in Generative AI.

    Find 3 important developments.

    Keep the answer concise.
    """,
    stream=True
)
```
